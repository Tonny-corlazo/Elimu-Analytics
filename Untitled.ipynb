```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("employees.csv", parse_dates=["hire_date"])

# light clean so plots look sane
df["name"] = df["name"].str.strip()
df["department"] = df["department"].str.lower().str.title()
df = df.drop_duplicates().reset_index(drop=True)
```

```python
df.head()
```

```python
sns.set_theme(style="whitegrid")
```

```python
sns.barplot(
    data=df,
    x="department",
    y="salary",
)
plt.title("Average Salary by Department")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.histplot(data=df, x="salary", hue="gender", bins=20, alpha=0.6)
```

```python
sns.boxplot(data=df, x="department", y="salary")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(data=df, x="department", y="salary", hue="gender")
```

```python
sns.scatterplot(data=df, x="years_exp", y="salary", hue="department")
plt.title("Experience vs Salary")
plt.show()
```

```python
sns.regplot(data=df, x="years_exp", y="salary", scatter_kws={"alpha":0.4})
```

```python
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True, cmap="viridis", fmt=".2f", center=0)
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.countplot(data=df, x="department", order=df["department"].value_counts().index)
plt.xticks(rotation=45)
plt.show()
```

```python
df["performance"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(df["salary"], bins=30, ax=axes[0, 0])
axes[0, 0].set_title("Salary")

sns.histplot(df["age"], bins=20, ax=axes[0, 1])
axes[0, 1].set_title("Age")

sns.histplot(df["years_exp"], bins=20, ax=axes[1, 0])
axes[1, 0].set_title("Years Exp")

sns.histplot(df["satisfaction"].dropna(), bins=10, ax=axes[1, 1])
axes[1, 1].set_title("Satisfaction")

plt.tight_layout()
plt.show()
```

```python
df.set_index("hire_date").resample("M").size().plot()
plt.title("Hires per Month")
plt.ylabel("Count")
plt.show()
```

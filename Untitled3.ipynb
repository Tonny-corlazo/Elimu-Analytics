```python
students_df["base_ability"] = np.random.normal(65, 12, 40).clip(30, 95)
students_df["growth"] = np.random.normal(0.8, 2.5, 40)
subjects = ["Mathematics", "Physics", "Chemistry", "Biology", "English", "Kiswahili"]
```

```python
for subj in subjects:
    students_df[f"bias_{subj}"] = np.random.uniform(0, 8, 40)

students_df["noise"] = np.random.uniform(3, 10, 40)
students_df.head()
```

```python
terms = [
    {"term": 1, "exam_type": "Opener",   "date": "2026-01-15"},
    {"term": 1, "exam_type": "End Term", "date": "2026-04-01"},
    {"term": 2, "exam_type": "Opener",   "date": "2026-05-10"},
    {"term": 2, "exam_type": "End Term", "date": "2026-07-28"},
    {"term": 3, "exam_type": "Opener",   "date": "2026-09-05"},
    {"term": 3, "exam_type": "End Term", "date": "2026-11-15"},
]
exams_df = pd.DataFrame(terms)
print(exams_df)
```

```python
rows = []

for _, student in students_df.iterrows():
    for exam_idx, exam in enumerate(terms):
        for subj in subjects:
            # Start from their base ability
            mark = student["base_ability"]
            
            # Add their subject bias
            mark += student[f"bias_{subj}"]
            
            # Add growth over time (each exam compounds the growth)
            mark += student["growth"] * exam_idx
            
            # Add random noise for this particular exam
            mark += np.random.normal(0, student["noise"])
            
            # Clip to realistic range
            mark = np.clip(mark, 5, 99)
            
            rows.append({
                "student_id": student["student_id"],
                "name":       student["name"],
                "class":      student["class"],
                "gender":     student["gender"],
                "term":       exam["term"],
                "exam_type":  exam["exam_type"],
                "exam_date":  exam["date"],
                "subject":    subj,
                "marks":      round(mark, 1),
                "out_of":     100,
            })

df = pd.DataFrame(rows)
print(f"Total rows: {len(df):,}")     # 40 × 6 × 6 = 1,440
print(df.head(12))
```

```python
print("Shape:", df.shape)
print("\nRows per student:", df.groupby("student_id").size().unique())
print("\nSubjects:", df["subject"].unique())
print("\nExams:", df["exam_type"].unique())
print("\nMarks range:", df["marks"].min(), "to", df["marks"].max())
print("\nMarks mean:", round(df["marks"].mean(), 2))

# Check for any missing values
print("\nNulls:", df.isna().sum().sum())
```

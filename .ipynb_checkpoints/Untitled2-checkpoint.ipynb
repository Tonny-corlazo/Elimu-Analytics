```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Simple Data
data = {
    'student': ['You', 'Friend1', 'Friend2', 'Friennd3', 'Friend4'],
    'hours_studied': [8, 5, 7, 3, 6],
    'exam_score': [88, 72, 85, 65, 78]
}

df = pd.DataFrame(data)
print("📊 Study Data")
print(df)
print(f"\nAverage hours studied: {df['hours_studied'].mean():.1f}")
print(f"Average Exam Score: {df['exam_score'].mean():.1f}")

# Correlation, More hours = higher score
corr = df['hours_studied'].corr(df['exam_score'])
print(f"\nCorrelation: {corr:.2f}")

plt.figure(figsize=(10, 5))
plt.scatter(df['hours_studied'], df['exam_score'], color='blue', s=100)
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.grid(True, alpha=0.3)
plt.show()
```

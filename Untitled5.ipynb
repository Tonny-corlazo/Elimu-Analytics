```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ─── 1. LOAD ───
df = pd.read_csv("elimu_analytics_2026.csv", parse_dates=["exam_date"])
df["percentage"] = (df["marks"] / df["out_of"] * 100).round(1)

def grade(pct):
    if pct >= 80: return "A"
    if pct >= 75: return "A-"
    if pct >= 70: return "B+"
    if pct >= 65: return "B"
    if pct >= 60: return "B-"
    if pct >= 55: return "C+"
    if pct >= 50: return "C"
    if pct >= 45: return "C-"
    if pct >= 40: return "D+"
    if pct >= 35: return "D"
    if pct >= 30: return "D-"
    return "E"

df["grade"] = df["percentage"].apply(grade)
df["exam_num"] = (df["term"] - 1) * 2 + df["exam_type"].map({"Opener": 1, "End Term": 2})

# ─── 2. THEME ───
sns.set_theme(
    context="notebook",
    style="darkgrid",
    palette="deep",
    font="sans-serif",
    font_scale=1,
    rc={"figure.facecolor": "#0e1117",
        "axes.facecolor": "#161b22",
        "axes.edgecolor": "white",
        "axes.labelcolor": "white",
        "xtick.color": "white",
        "ytick.color": "white",
        "text.color": "white",
        "grid.color": "#30363d"}
)

# ─── 3. FIGURE ───
fig, axes = plt.subplots(3, 3, figsize=(26, 16))
fig.suptitle(
    "Class 3A — Academic Performance Report 2026",
    fontsize=16, weight="bold", color="white", x=0.05, ha="left"
)
overall_mean = df["percentage"].mean()

# ─── PANEL 1: Subject averages ───
subj_order = df.groupby("subject")["percentage"].mean().sort_values().index
sns.barplot(data=df, x="subject", y="percentage",
            order=subj_order, ax=axes[0, 0], errorbar=None,
            palette="viridis")
axes[0, 0].axhline(overall_mean, color="red", linestyle="--",
                   label=f"Class mean: {overall_mean:.1f}%")
axes[0, 0].set_title("Average Performance by Subject", weight="bold")
axes[0, 0].set_xlabel("")
axes[0, 0].set_ylabel("Mean %")
axes[0, 0].tick_params(axis="x", rotation=30)
axes[0, 0].legend(facecolor="#0e1117", labelcolor="white", fontsize=9)

# ─── PANEL 2: Exam progression ───
class_prog = df.groupby("exam_num")["percentage"].mean()
sns.lineplot(data=df, x="exam_num", y="percentage",
             estimator="mean", marker="o", linewidth=2.5,
             ax=axes[0, 1], color="#4ade80")
axes[0, 1].set_title("Class Mean Across the Year", weight="bold")
axes[0, 1].set_xlabel("Exam (1 = Term 1 Opener, 6 = Term 3 End Term)")
axes[0, 1].set_ylabel("Mean %")
axes[0, 1].set_xticks(range(1, 7))

# ─── PANEL 3: Grade distribution ───
grade_order = ["A","A-","B+","B","B-","C+","C","C-","D+","D","D-","E"]
grade_counts = df["grade"].value_counts().reindex(grade_order, fill_value=0)
bars = axes[0, 2].bar(grade_counts.index, grade_counts.values,
                       color=sns.color_palette("RdYlGn_r", len(grade_counts)))
axes[0, 2].set_title("Grade Distribution (all exams)", weight="bold")
axes[0, 2].set_xlabel("Grade")
axes[0, 2].set_ylabel("Count")
for bar, count in zip(bars, grade_counts.values):
    if count > 0:
        axes[0, 2].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
                        str(count), ha="center", color="white", fontsize=8)

# ─── PANEL 4: Gender gap by subject ───
gender_subject = df.pivot_table(index="subject", columns="gender",
                                values="percentage", aggfunc="mean")
print(gender_subject)

gender_subject = gender_subject.loc[subj_order]
gender_subject.plot(kind="barh", ax=axes[1, 0],
                     color=["#4da6ff", "#ff6b9d"], width=0.75)
axes[1, 0].set_title("Gender Gap by Subject", weight="bold")
axes[1, 0].set_xlabel("Mean %")
axes[1, 0].set_ylabel("")
axes[1, 0].legend(title="Gender", facecolor="#0e1117",
                   labelcolor="white")

# ─── PANEL 5: Top 10 students ───
top10 = (df.groupby(["student_id", "name"])["percentage"]
           .mean().sort_values(ascending=False).head(10).reset_index())
sns.barplot(data=top10, y="name", x="percentage",
            ax=axes[1, 1], palette="Greens_r")
axes[1, 1].set_title("Top 10 Students (overall mean)", weight="bold")
axes[1, 1].set_xlabel("Mean %")
axes[1, 1].set_ylabel("")
axes[1, 1].axvline(overall_mean, color="red", linestyle="--", alpha=0.6)

# ─── PANEL 6: At-risk students ───
recent = df[df["exam_num"].isin([5, 6])]
at_risk = (recent.groupby(["student_id", "name"])["percentage"]
              .mean().sort_values().head(10).reset_index())
sns.barplot(data=at_risk, y="name", x="percentage",
            ax=axes[1, 2], palette="Reds")
axes[1, 2].set_title(" At-Risk Students (recent exams)", weight="bold")
axes[1, 2].set_xlabel("Mean %")
axes[1, 2].set_ylabel("")
axes[1, 2].axvline(50, color="yellow", linestyle="--",
                    label="50% threshold")
axes[1, 2].legend(facecolor="#0e1117", labelcolor="white")


# ─── PANEL 7: Relationship btwn openers and end terms exams in evry term  ───
sns.barplot(
    data=df,
    x="term",
    y="percentage",
    hue="exam_type",
    ax=axes[2, 0],
    errorbar=None,
    palette={"Opener": "#4da6ff", "End Term": "#ffcc00"}
)

axes[2, 0].set_title("Class Mean: Opener vs End Term per Term", weight="bold")
axes[2, 0].set_xlabel("Term")
axes[2, 0].set_ylabel("Mean %")
axes[2, 0].legend(title="Exam", facecolor="#0e1117", labelcolor="white")

sns.histplot(
    data=df,
    x="percentage",
    bins=30,
    kde=True,
    ax=axes[2, 1],
    color="#4da6ff",
    alpha=0.6
)

sns.rugplot(
    data=df,
    x="percentage",
    ax=axes[2, 1],
    color="#ffcc00",
    height=0.1
)

axes[2, 1].set_title("Percentage Distribution", weight="bold")
axes[2, 1].set_xlabel("Percentage")
axes[2, 1].set_ylabel("Count")

plt.tight_layout()
plt.savefig("class_3A_report.png", dpi=150,
            facecolor="#0e1117", bbox_inches="tight")
plt.show()
```

```python
from statsbombpy import sb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mplsoccer import Pitch

# ────────────────────────────────────────────────
# 1. LOAD DATA
# ────────────────────────────────────────────────
matches = sb.matches(competition_id=55, season_id=282)
match = matches[
    (matches['home_team'] == 'Spain') &
    (matches['away_team'] == 'Italy')
].iloc[0]

events = sb.events(match_id=match['match_id'])
events = events.sort_values('index').reset_index(drop=True)

# ────────────────────────────────────────────────
# 2. FILTER SLICES
# ────────────────────────────────────────────────
shots_all   = events[events['type'] == 'Shot']
passes_spain = events[(events['type'] == 'Pass') & (events['team'] == 'Spain')]
passes_italy = events[(events['type'] == 'Pass') & (events['team'] == 'Italy')]

# ────────────────────────────────────────────────
# 3. FIGURE + GRID
# ────────────────────────────────────────────────
plt.style.use('dark_background')
fig = plt.figure(figsize=(18, 10), facecolor='#0e1117')
gs = fig.add_gridspec(2, 3, hspace=0.25, wspace=0.15,
                      left=0.03, right=0.97, top=0.92, bottom=0.05)

# Create six axes
ax_shots  = fig.add_subplot(gs[0, 0])
ax_moment = fig.add_subplot(gs[0, 1])
ax_heat   = fig.add_subplot(gs[0, 2])
ax_pass_s = fig.add_subplot(gs[1, 0])
ax_stats  = fig.add_subplot(gs[1, 1])
ax_pass_i = fig.add_subplot(gs[1, 2])

fig.suptitle(
    f"{match['home_team']} {match['home_score']} - {match['away_score']} {match['away_team']}  |  Euro 2024",
    color='white', fontsize=16, weight='bold', x=0.03, ha='left'
)

# ────────────────────────────────────────────────
# 4. PANEL 1 — SHOT MAP (both teams)
# ────────────────────────────────────────────────
pitch1 = Pitch(pitch_type='statsbomb', pitch_color='#161b22', line_color='white')
pitch1.draw(ax=ax_shots)

for _, s in shots_all.iterrows():
    x, y = s['location']
    xg = s['shot_statsbomb_xg']
    color = '#ffcc00' if s['team'] == 'Spain' else '#4da6ff'
    if s['shot_outcome'] == 'Goal':
        color = '#ff3b3b'    # goals always red
    pitch1.scatter(x, y, s=xg * 2000, ax=ax_shots,
                   color=color, edgecolors='white',
                   alpha=0.85, zorder=3)

ax_shots.set_title("Shots — sized by xG", color='white', fontsize=11, loc='left')

# ────────────────────────────────────────────────
# 5. PANEL 2 — MOMENTUM (rolling xG)
# ────────────────────────────────────────────────
shots_all = shots_all.copy()
shots_all['minute'] = shots_all['minute'].astype(int)

spain_xg = shots_all[shots_all['team'] == 'Spain'].groupby('minute')['shot_statsbomb_xg'].sum()
italy_xg = shots_all[shots_all['team'] == 'Italy'].groupby('minute')['shot_statsbomb_xg'].sum()

minutes = np.arange(0, 95)
spain_series = spain_xg.reindex(minutes, fill_value=0).rolling(10, min_periods=1).sum()
italy_series = italy_xg.reindex(minutes, fill_value=0).rolling(10, min_periods=1).sum()

ax_moment.fill_between(minutes, 0, spain_series, color='#ffcc00', alpha=0.6, label='Spain')
ax_moment.fill_between(minutes, 0, -italy_series, color='#4da6ff', alpha=0.6, label='Italy')
ax_moment.axhline(0, color='white', linewidth=0.5)
ax_moment.set_facecolor('#161b22')
ax_moment.set_title("Momentum (10-min rolling xG)", color='white', fontsize=11, loc='left')
ax_moment.set_xlabel("Minute", color='white')
ax_moment.tick_params(colors='white')
ax_moment.legend(loc='upper right', facecolor='#0e1117', labelcolor='white')

# ────────────────────────────────────────────────
# 6. PANEL 3 — HEATMAP (Spain pass origins)
# ────────────────────────────────────────────────
pitch3 = Pitch(pitch_type='statsbomb', pitch_color='#161b22', line_color='white')
pitch3.draw(ax=ax_heat)

xs = passes_spain['location'].apply(lambda loc: loc[0])
ys = passes_spain['location'].apply(lambda loc: loc[1])
pitch3.hexbin(xs, ys, ax=ax_heat, gridsize=(20, 12),
              cmap='YlOrRd', alpha=0.75, zorder=2)
ax_heat.set_title("Spain Pass Heatmap", color='white', fontsize=11, loc='left')

# ────────────────────────────────────────────────
# 7. PANEL 4 — SPAIN PASS MAP
# ────────────────────────────────────────────────
pitch4 = Pitch(pitch_type='statsbomb', pitch_color='#161b22', line_color='white')
pitch4.draw(ax=ax_pass_s)

for _, p in passes_spain.iterrows():
    x, y = p['location']
    ex, ey = p['pass_end_location']
    pitch4.arrows(x, y, ex, ey, ax=ax_pass_s, width=1.5,
                  headwidth=3, headlength=4,
                  color='#ffcc00', alpha=0.35)
ax_pass_s.set_title("Spain Pass Map", color='white', fontsize=11, loc='left')

# ────────────────────────────────────────────────
# 8. PANEL 5 — STAT BARS
# ────────────────────────────────────────────────
stats = {
    "Goals":        [match['home_score'], match['away_score']],
    "Shots":        [len(shots_all[shots_all['team']=='Spain']),
                     len(shots_all[shots_all['team']=='Italy'])],
    "Passes":       [len(passes_spain), len(passes_italy)],
    "xG (x10)":     [round(shots_all[shots_all['team']=='Spain']['shot_statsbomb_xg'].sum()*10, 1),
                     round(shots_all[shots_all['team']=='Italy']['shot_statsbomb_xg'].sum()*10, 1)],
}
labels = list(stats.keys())
spain_vals = [v[0] for v in stats.values()]
italy_vals = [v[1] for v in stats.values()]

y_pos = np.arange(len(labels))
ax_stats.barh(y_pos - 0.2, spain_vals, height=0.4, color='#ffcc00', label='Spain')
ax_stats.barh(y_pos + 0.2, italy_vals, height=0.4, color='#4da6ff', label='Italy')
ax_stats.set_yticks(y_pos)
ax_stats.set_yticklabels(labels, color='white')
ax_stats.invert_yaxis()
ax_stats.set_facecolor('#161b22')
ax_stats.set_title("Match Stats", color='white', fontsize=11, loc='left')
ax_stats.tick_params(colors='white')
ax_stats.legend(facecolor='#0e1117', labelcolor='white')

# ────────────────────────────────────────────────
# 9. PANEL 6 — ITALY PASS MAP
# ────────────────────────────────────────────────
pitch6 = Pitch(pitch_type='statsbomb', pitch_color='#161b22', line_color='white')
pitch6.draw(ax=ax_pass_i)

for _, p in passes_italy.iterrows():
    x, y = p['location']
    ex, ey = p['pass_end_location']
    pitch6.arrows(x, y, ex, ey, ax=ax_pass_i, width=1.5,
                  headwidth=3, headlength=4,
                  color='#4da6ff', alpha=0.35)
ax_pass_i.set_title("Italy Pass Map", color='white', fontsize=11, loc='left')

# ────────────────────────────────────────────────
# 10. SAVE + SHOW
# ────────────────────────────────────────────────
plt.savefig("match_report.png", dpi=150, facecolor='#0e1117', bbox_inches='tight')
plt.show()
```
